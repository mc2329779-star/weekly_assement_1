```python
income = float(input("Enter annual income: ₹"))

tax = 0

print("\nTax Breakdown:")

if income <= 300000:
    print("Up to ₹3,00,000: ₹0.00")

else:
    print("Up to ₹3,00,000: ₹0.00")

    if income > 300000:
        amount = min(income, 700000) - 300000
        slab_tax = amount * 0.05
        tax += slab_tax
        print(f"₹3,00,001 - ₹7,00,000: ₹{slab_tax:.2f}")

    if income > 700000:
        amount = min(income, 1000000) - 700000
        slab_tax = amount * 0.10
        tax += slab_tax
        print(f"₹7,00,001 - ₹10,00,000: ₹{slab_tax:.2f}")

    if income > 1000000:
        amount = min(income, 1200000) - 1000000
        slab_tax = amount * 0.15
        tax += slab_tax
        print(f"₹10,00,001 - ₹12,00,000: ₹{slab_tax:.2f}")

    if income > 1200000:
        amount = min(income, 1500000) - 1200000
        slab_tax = amount * 0.20
        tax += slab_tax
        print(f"₹12,00,001 - ₹15,00,000: ₹{slab_tax:.2f}")

    if income > 1500000:
        amount = income - 1500000
        slab_tax = amount * 0.30
        tax += slab_tax
        print(f"Above ₹15,00,000: ₹{slab_tax:.2f}")

print("-----------------------------")
print(f"Total Tax: ₹{tax:.2f}")
```

```python
armstrong = []

for num in range(1, 10000):
    temp = num
    digits = 0

    # Count number of digits
    while temp > 0:
        digits += 1
        temp //= 10

    # Calculate sum of digits raised to the power of digits
    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    if total == num:
        armstrong.append(num)

print("Armstrong numbers:")
print(", ".join(map(str, armstrong)))
print("Count:", len(armstrong))
```

```python
n = int(input("Enter N: "))

def is_prime(num):
    if num < 2:
        return False

    i = 2
    while i * i <= num:   # checks only up to √num
        if num % i == 0:
            return False
        i += 1

    return True


if is_prime(n):
    print(n, "is Prime")
else:
    print(n, "is Not Prime")

prime_sum = 0

for num in range(2, n + 1):
    if is_prime(num):
        prime_sum += num

print("Sum of primes from 2 to", n, "=", prime_sum)
```

```python
num = int(input("Enter a number: "))

original = num

if num < 0:
    print("Reversed:", -num)
    print("Not Palindrome")
else:
    reverse = 0
    temp = num

    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp //= 10

    print("Reversed:", reverse)

    if original == reverse:
        print("Palindrome")
    else:
        print("Not Palindrome")
```

```python
a = int(input("Enter first positive integer: "))
b = int(input("Enter second positive integer: "))

x = a
y = b

# Euclidean algorithm for HCF
while y != 0:
    x, y = y, x % y

hcf = x

# HCF × LCM = a × b
lcm = (a * b) // hcf

print("HCF:", hcf)
print("LCM:", lcm)
```

```python
sentence = input("Enter a sentence: ")

total_characters = len(sentence)
words = 0
vowels = 0
consonants = 0
digits = 0
spaces = 0
special_characters = 0

in_word = False

for ch in sentence:   # Single pass
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

        if not in_word:
            words += 1
            in_word = True

    elif ch.isdigit():
        digits += 1
        if not in_word:
            words += 1
            in_word = True

    elif ch.isspace():
        spaces += 1
        in_word = False

    else:
        special_characters += 1

print("\n----- String Scanner -----")
print("Total characters:", total_characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special_characters)
```

```python
marks = list(map(int, input("Enter 10 marks: ").split()))

maximum = marks[0]
minimum = marks[0]
total = 0

for mark in marks:
    if mark > maximum:
        maximum = mark

    if mark < minimum:
        minimum = mark

    total += mark

average = total / len(marks)

above_average = 0
second_highest = None

for mark in marks:
    if mark > average:
        above_average += 1

    if mark != maximum:
        if second_highest is None or mark > second_highest:
            second_highest = mark

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Average:", average)
print("Students above average:", above_average)
print("Second highest:", second_highest)
```

```python
review = input("Enter Swiggy review: ")

words = review.lower().split()

frequency = {}

for word in words:
    word = word.strip(".,!?;:")

    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Sort words by frequency in descending order
sorted_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

print("\nWord Frequency:")
for word, count in sorted_words:
    print(word, ":", count)

# Most frequent word
most_frequent_word, highest_count = sorted_words[0]

print("\nMost frequent word:", most_frequent_word)
print("Count:", highest_count)
```

```python
section_a = {"Python", "Java", "C++", "SQL"}
section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}

# (a) Languages known by BOTH
both = section_a & section_b

# (b) Languages ONLY by Section A
only_a = section_a - section_b

# (c) Languages ONLY by Section B
only_b = section_b - section_a

# (d) Total unique languages
total_unique = section_a | section_b

# (e) Languages exclusive to one section
exclusive = section_a ^ section_b

print("Both sections:", both)
print("Only Section A:", only_a)
print("Only Section B:", only_b)
print("Total unique languages:", len(total_unique))
print("Exclusive languages:", exclusive)
```

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)

days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

# (a) Maximum price and day
max_price = max(prices)
max_day = days[prices.index(max_price)]

# (b) Minimum price and day
min_price = min(prices)
min_day = days[prices.index(min_price)]

# (c) Average price
average = sum(prices) / len(prices)

# (d) Biggest single-day gain
biggest_gain = prices[1] - prices[0]
gain_from = days[0]
gain_to = days[1]

for i in range(1, len(prices)):
    gain = prices[i] - prices[i - 1]

    if gain > biggest_gain:
        biggest_gain = gain
        gain_from = days[i - 1]
        gain_to = days[i]

# (e) Total volatility
volatility = 0

for i in range(1, len(prices)):
    volatility += abs(prices[i] - prices[i - 1])

print("Maximum price:", max_price, "on", max_day)
print("Minimum price:", min_price, "on", min_day)
print("Average price: ₹{:.2f}".format(average))
print("Biggest single-day gain: ₹{} ({} to {})".format(
    biggest_gain, gain_from, gain_to
))
print("Total volatility: ₹", volatility)
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount
    else:
        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = base_amount + gst_amount

    return (
        round(base_amount, 2),
        round(gst_amount, 2),
        round(total_amount, 2)
    )


print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

```python
def power(base, exp):
    if exp == 0:
        return 1

    if exp < 0:
        return 1 / power(base, -exp)

    return base * power(base, exp - 1)


print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
import numpy as np

# (a) Create 5x5 identity matrix
matrix = np.eye(5)

print("Identity Matrix:")
print(matrix)

# (b) Multiply every element by 10
matrix = matrix * 10

print("\nAfter multiplying by 10:")
print(matrix)

# (c) Transpose the matrix
# An identity matrix is symmetric, so its transpose is itself.
transpose = matrix.T

print("\nTranspose:")
print(transpose)

# (d) Sum of diagonal elements
diagonal_sum = np.trace(matrix)

print("\nSum of diagonal elements:", diagonal_sum)

# (e) Sum of all elements
total_sum = np.sum(matrix)

print("Sum of all elements:", total_sum)
```

```python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# (a) Total weekly sales
total_sales = np.sum(sales)

# (b) Daily average
average_sales = np.mean(sales)

# (c) Day with maximum sales
max_day = days[np.argmax(sales)]

# (d) Number of days above weekly average
above_average = np.sum(sales > average_sales)

# (e) Percentage growth from Monday to Sunday
growth = ((sales[-1] - sales[0]) / sales[0]) * 100

# (f) Day-on-day percentage changes
day_on_day = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("Total weekly sales: ₹", total_sales)
print("Daily average: ₹{:.2f}".format(average_sales))
print("Maximum sales day:", max_day)
print("Days above average:", above_average)
print("Monday to Sunday growth: {:.2f}%".format(growth))
print("Day-on-day % changes:", np.round(day_on_day, 2))
```
